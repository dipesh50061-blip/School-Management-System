```python
import json
import os
from datetime import date

FILE_NAME = "students.json"


# -------------------- File Handling --------------------

def load_data():
    if os.path.exists(FILE_NAME):
        with open(FILE_NAME, "r") as file:
            return json.load(file)
    return {}


def save_data(data):
    with open(FILE_NAME, "w") as file:
        json.dump(data, file, indent=4)


# -------------------- Add Student --------------------

def add_student(data):
    roll = input("Enter Roll Number: ")

    if roll in data:
        print("Student already exists!")
        return

    name = input("Enter Name: ")
    age = int(input("Enter Age: "))
    marks = float(input("Enter Marks: "))

    data[roll] = {
        "name": name,
        "age": age,
        "marks": marks,
        "attendance": {}
    }

    save_data(data)
    print("Student added successfully!")


# -------------------- View All Students --------------------

def view_students(data):
    if not data:
        print("No students found.")
        return

    for roll, info in data.items():
        attendance = info["attendance"]

        present = sum(
            1 for status in attendance.values() if status == "Present"
        )
        absent = sum(
            1 for status in attendance.values() if status == "Absent"
        )

        total = present + absent

        print("-" * 40)
        print(f"Roll No : {roll}")
        print(f"Name    : {info['name']}")
        print(f"Age     : {info['age']}")
        print(f"Marks   : {info['marks']}")
        print(f"Present : {present}")
        print(f"Absent  : {absent}")
        print(f"Total   : {total}")


# -------------------- Search Student --------------------

def search_student(data):
    roll = input("Enter Roll Number: ")

    if roll not in data:
        print("Student not found.")
        return

    info = data[roll]
    attendance = info["attendance"]

    present = sum(
        1 for status in attendance.values() if status == "Present"
    )
    absent = sum(
        1 for status in attendance.values() if status == "Absent"
    )

    print("-" * 40)
    print(f"Roll No : {roll}")
    print(f"Name    : {info['name']}")
    print(f"Age     : {info['age']}")
    print(f"Marks   : {info['marks']}")
    print(f"Present : {present}")
    print(f"Absent  : {absent}")


# -------------------- Update Marks --------------------

def update_marks(data):
    roll = input("Enter Roll Number: ")

    if roll not in data:
        print("Student not found.")
        return

    marks = float(input("Enter New Marks: "))
    data[roll]["marks"] = marks

    save_data(data)
    print("Marks updated successfully!")


# -------------------- Delete Student --------------------

def delete_student(data):
    roll = input("Enter Roll Number: ")

    if roll not in data:
        print("Student not found.")
        return

    del data[roll]
    save_data(data)

    print("Student deleted successfully!")


# -------------------- Show Topper --------------------

def topper(data):
    if not data:
        print("No students available.")
        return

    top = max(data.items(), key=lambda x: x[1]["marks"])

    print("\n===== TOPPER =====")
    print(f"Roll No : {top[0]}")
    print(f"Name    : {top[1]['name']}")
    print(f"Marks   : {top[1]['marks']}")


# -------------------- Mark Today's Attendance --------------------

def mark_attendance(data):
    roll = input("Enter Roll Number: ")

    if roll not in data:
        print("Student not found.")
        return

    today = str(date.today())

    if today in data[roll]["attendance"]:
        print("Attendance already marked for today.")
        return

    status = input("Enter P for Present or A for Absent: ").upper()

    if status == "P":
        data[roll]["attendance"][today] = "Present"
    elif status == "A":
        data[roll]["attendance"][today] = "Absent"
    else:
        print("Invalid input.")
        return

    save_data(data)
    print("Attendance saved successfully!")


# -------------------- View Attendance History --------------------

def view_attendance(data):
    roll = input("Enter Roll Number: ")

    if roll not in data:
        print("Student not found.")
        return

    attendance = data[roll]["attendance"]

    if not attendance:
        print("No attendance records.")
        return

    print("\n===== Attendance History =====")

    for day, status in sorted(attendance.items()):
        print(f"{day} : {status}")

    present = sum(
        1 for status in attendance.values() if status == "Present"
    )
    absent = sum(
        1 for status in attendance.values() if status == "Absent"
    )

    total = present + absent

    if total == 0:
        percentage = 0
    else:
        percentage = (present / total) * 100

    print("\n===== Summary =====")
    print(f"Present              : {present}")
    print(f"Absent               : {absent}")
    print(f"Total Classes        : {total}")
    print(f"Attendance Percentage: {percentage:.2f}%")


# -------------------- Main Menu --------------------

def main():
    data = load_data()

    while True:
        print("\n========== STUDENT MANAGEMENT SYSTEM ==========")
        print("1. Add Student")
        print("2. View Students")
        print("3. Search Student")
        print("4. Update Marks")
        print("5. Delete Student")
        print("6. Show Topper")
        print("7. Mark Today's Attendance")
        print("8. View Attendance History")
        print("9. Exit")

        choice = input("\nEnter your choice: ")

        try:
            if choice == "1":
                add_student(data)

            elif choice == "2":
                view_students(data)

            elif choice == "3":
                search_student(data)

            elif choice == "4":
                update_marks(data)

            elif choice == "5":
                delete_student(data)

            elif choice == "6":
                topper(data)

            elif choice == "7":
                mark_attendance(data)

            elif choice == "8":
                view_attendance(data)

            elif choice == "9":
                print("Thank you!")
                break

            else:
                print("Invalid choice.")

        except ValueError:
            print("Please enter valid numeric input.")


if __name__ == "__main__":
    main()
```

```python

"""
School Management System
Features:
- Auto-generated Student IDs
- CRUD students
- Class/Section
- Subject-wise marks
- Total/Average/Percentage/Grade/Pass-Fail
- Daily attendance (date-wise)
- Bulk attendance
- Attendance history/date range
- Edit attendance
- Low attendance alerts
- Sorting/filtering
- Result reports
Stores data in school_db.json
"""

import json, os
from datetime import datetime

DB="school_db.json"
SUBJECTS=["Python","Mathematics","Statistics","English","AI Fundamentals"]

def load():
    if os.path.exists(DB):
        with open(DB,"r") as f:
            return json.load(f)
    return {"counter":1,"students":{}}

def save(db):
    with open(DB,"w") as f:
        json.dump(db,f,indent=2)

def gen_id(db):
    sid=f"STU{db['counter']:03d}"
    db["counter"]+=1
    return sid

def metrics(stu):
    vals=list(stu["marks"].values())
    total=sum(vals)
    avg=total/len(SUBJECTS)
    pct=avg
    if pct>=90:g="A+"
    elif pct>=80:g="A"
    elif pct>=70:g="B"
    elif pct>=60:g="C"
    elif pct>=50:g="D"
    else:g="F"
    status="PASS" if all(v>=35 for v in vals) else "FAIL"
    att=stu["attendance"]
    p=sum(v=="Present" for v in att.values())
    a=sum(v=="Absent" for v in att.values())
    perc=100*p/(p+a) if p+a else 0
    return total,avg,pct,g,status,p,a,perc

def add_student(db):
    sid=gen_id(db)
    name=input("Name: ")
    age=int(input("Age: "))
    cls=input("Class: ")
    sec=input("Section: ")
    marks={}
    for s in SUBJECTS:
        marks[s]=float(input(f"{s}: "))
    db["students"][sid]={
        "name":name,"age":age,"class":cls,"section":sec,
        "marks":marks,"attendance":{}
    }
    save(db)
    print("Created",sid)

def update_student(db):
    sid=input("ID: ")
    s=db["students"].get(sid)
    if not s:
        print("Not found");return
    n=input(f"Name({s['name']}): ") or s["name"]
    a=input(f"Age({s['age']}): ")
    c=input(f"Class({s['class']}): ") or s["class"]
    sec=input(f"Section({s['section']}): ") or s["section"]
    s["name"]=n
    if a:s["age"]=int(a)
    s["class"]=c
    s["section"]=sec
    save(db)

def update_marks(db):
    sid=input("ID: ")
    s=db["students"].get(sid)
    if not s:return print("Not found")
    for sub in SUBJECTS:
        val=input(f"{sub} ({s['marks'][sub]}): ")
        if val:
            s["marks"][sub]=float(val)
    save(db)

def mark_bulk(db):
    d=input("Date YYYY-MM-DD(blank=today): ") or datetime.now().strftime("%Y-%m-%d")
    for sid,s in db["students"].items():
        if d in s["attendance"]:
            continue
        st=input(f"{sid} {s['name']} P/A: ").upper()
        s["attendance"][d]="Present" if st=="P" else "Absent"
    save(db)

def edit_att(db):
    sid=input("ID: "); d=input("Date YYYY-MM-DD: ")
    s=db["students"].get(sid)
    if not s:return
    print("Current:",s["attendance"].get(d))
    st=input("P/A: ").upper()
    s["attendance"][d]="Present" if st=="P" else "Absent"
    save(db)

def attendance_range(db):
    sid=input("ID: ")
    s=db["students"].get(sid)
    if not s:return
    start=input("Start YYYY-MM-DD: ")
    end=input("End YYYY-MM-DD: ")
    for d,v in sorted(s["attendance"].items()):
        if start<=d<=end:
            print(d,v)

def report(db):
    rows=[]
    for sid,s in db["students"].items():
        total,avg,pct,g,stat,p,a,att=metrics(s)
        alert="LOW" if att<75 else ""
        rows.append((sid,s["name"],total,pct,g,stat,att,alert))
    rows.sort(key=lambda x:x[2],reverse=True)
    print("ID | Name | Total | % | Grade | Status | Attend% | Alert")
    for r in rows:
        print(*r,sep=" | ")

def view(db):
    sort=input("Sort(name/marks/attendance/none): ")
    items=list(db["students"].items())
    if sort=="name":
        items.sort(key=lambda x:x[1]["name"])
    elif sort=="marks":
        items.sort(key=lambda x:metrics(x[1])[0],reverse=True)
    elif sort=="attendance":
        items.sort(key=lambda x:metrics(x[1])[7],reverse=True)
    flt=input("Filter class(blank none): ")
    for sid,s in items:
        if flt and s["class"]!=flt: continue
        total,avg,pct,g,stat,p,a,att=metrics(s)
        print("-"*40)
        print(sid,s["name"],f"Class {s['class']}-{s['section']}")
        print("Total",total,"Avg",round(avg,2),"Grade",g,stat)
        print("Attendance%",round(att,2),("LOW" if att<75 else ""))

def delete(db):
    sid=input("ID: ")
    db["students"].pop(sid,None)
    save(db)

def main():
    db=load()
    menu="""
1 Add Student
2 Update Student
3 Update Marks
4 View/Sort/Filter
5 Bulk Attendance
6 Attendance by Date Range
7 Edit Attendance
8 Result Report
9 Delete Student
0 Exit
"""
    while True:
        print(menu)
        ch=input("Choice: ")
        if ch=="1": add_student(db)
        elif ch=="2": update_student(db)
        elif ch=="3": update_marks(db)
        elif ch=="4": view(db)
        elif ch=="5": mark_bulk(db)
        elif ch=="6": attendance_range(db)
        elif ch=="7": edit_att(db)
        elif ch=="8": report(db)
        elif ch=="9": delete(db)
        elif ch=="0":
            save(db);break

if __name__=="__main__":
    main()
```

```python
"""
School Management System v2.0
Features:
- PIN-based Login Security
- Auto-generated Student IDs
- CRUD Students
- Class/Section Management
- Custom Subjects (Global & Class-wise)
- Subject-wise Marks
- Total/Average/Percentage/Grade/Pass-Fail
- Daily Attendance (date-wise)
- Bulk Attendance
- Attendance History/Date Range
- Edit Attendance
- Low Attendance Alerts
- Sorting/Filtering
- Result Reports
- Fee Management (Structure + Payments)
- Class Timetable
- Notice Board
- Student ID Card Generator
- Data Export/Import (CSV + JSON)
- Analytics & Reports
"""

import json
import os
import csv
import hashlib
from datetime import datetime, timedelta
from collections import defaultdict
import uuid

# ==================== DATABASE ====================
DB = "school_data.json"

def load_db():
    """Load database from file"""
    if os.path.exists(DB):
        with open(DB, "r") as f:
            return json.load(f)
    return {
        "pin_hash": "",
        "counter": 1,
        "students": {},
        "subjects": {
            "global": ["Mathematics", "English", "Science", "Hindi", "Social Studies"],
            "classwise": {}
        },
        "attendance": {},
        "results": {},
        "fees": {
            "structure": {},
            "payments": []
        },
        "timetable": {},
        "notices": []
    }

def save_db(db):
    """Save database to file"""
    with open(DB, "w") as f:
        json.dump(db, f, indent=2)

def hash_pin(pin):
    """Hash PIN for security"""
    return hashlib.sha256(pin.encode()).hexdigest()

# ==================== LOGIN SYSTEM ====================
def setup_pin(db):
    """Setup initial PIN"""
    print("\n" + "="*50)
    print("   🔐 SECURITY SETUP")
    print("="*50)
    
    while True:
        pin = input("\n   Set your 4-digit PIN: ").strip()
        if len(pin) == 4 and pin.isdigit():
            confirm = input("   Confirm PIN: ").strip()
            if pin == confirm:
                db["pin_hash"] = hash_pin(pin)
                save_db(db)
                print("\n   ✅ PIN set successfully!")
                return True
            else:
                print("   ❌ PINs don't match! Try again.")
        else:
            print("   ❌ PIN must be 4 digits!")

def login(db):
    """Login with PIN"""
    if not db.get("pin_hash"):
        return setup_pin(db)
    
    print("\n" + "="*50)
    print("   🔐 LOGIN")
    print("="*50)
    
    attempts = 3
    while attempts > 0:
        pin = input(f"\n   Enter PIN ({attempts} attempts left): ").strip()
        if hash_pin(pin) == db["pin_hash"]:
            print("\n   ✅ Login successful!")
            return True
        attempts -= 1
        if attempts > 0:
            print("   ❌ Invalid PIN!")
    
    print("\n   🚫 Too many failed attempts. Exiting...")
    return False

def change_pin(db):
    """Change existing PIN"""
    print("\n" + "-"*40)
    print("   CHANGE PIN")
    print("-"*40)
    
    old_pin = input("   Current PIN: ").strip()
    if hash_pin(old_pin) != db["pin_hash"]:
        print("   ❌ Invalid current PIN!")
        return
    
    new_pin = input("   New 4-digit PIN: ").strip()
    if len(new_pin) != 4 or not new_pin.isdigit():
        print("   ❌ PIN must be 4 digits!")
        return
    
    confirm = input("   Confirm new PIN: ").strip()
    if new_pin == confirm:
        db["pin_hash"] = hash_pin(new_pin)
        save_db(db)
        print("   ✅ PIN changed successfully!")
    else:
        print("   ❌ PINs don't match!")

# ==================== UTILITY FUNCTIONS ====================
def generate_student_id(db):
    """Generate unique student ID"""
    sid = f"STU{db['counter']:04d}"
    db["counter"] += 1
    return sid

def get_subjects_for_class(db, cls):
    """Get all subjects for a class"""
    subjects = db["subjects"]["global"].copy()
    if cls in db["subjects"]["classwise"]:
        subjects.extend(db["subjects"]["classwise"][cls])
    return subjects

def calculate_age(dob_str):
    """Calculate age from date of birth"""
    if not dob_str:
        return "N/A"
    try:
        dob = datetime.strptime(dob_str, "%Y-%m-%d")
        today = datetime.now()
        age = today.year - dob.year
        if today.month < dob.month or (today.month == dob.month and today.day < dob.day):
            age -= 1
        return age
    except:
        return "N/A"

def get_grade(percentage):
    """Get grade based on percentage"""
    if percentage >= 90: return "A+"
    elif percentage >= 80: return "A"
    elif percentage >= 70: return "B"
    elif percentage >= 60: return "C"
    elif percentage >= 50: return "D"
    else: return "F"

def get_attendance_percentage(db, student_id):
    """Calculate attendance percentage for a student"""
    att = db["attendance"].get(student_id, {})
    if not att:
        return 0
    present = sum(1 for v in att.values() if v == "Present")
    total = len(att)
    return (present / total * 100) if total > 0 else 0

# ==================== STUDENT MANAGEMENT ====================
def add_student(db):
    """Add new student"""
    print("\n" + "="*50)
    print("   👨‍🎓 ADD NEW STUDENT")
    print("="*50)
    
    sid = generate_student_id(db)
    name = input("\n   Student Name: ").strip()
    if not name:
        print("   ❌ Name is required!")
        db["counter"] -= 1
        return
    
    dob = input("   Date of Birth (YYYY-MM-DD): ").strip()
    
    print("\n   Available Classes: 1-12")
    cls = input("   Class: ").strip()
    if not cls:
        cls = "1"
    
    section = input("   Section (A/B/C): ").strip() or "A"
    
    father_name = input("   Father's Name: ").strip()
    mother_name = input("   Mother's Name: ").strip()
    phone = input("   Phone Number: ").strip()
    address = input("   Address: ").strip()
    
    # Initialize student
    subjects = get_subjects_for_class(db, cls)
    marks = {sub: 0 for sub in subjects}
    
    db["students"][sid] = {
        "id": sid,
        "name": name,
        "dob": dob,
        "class": cls,
        "section": section,
        "father_name": father_name,
        "mother_name": mother_name,
        "phone": phone,
        "address": address,
        "marks": marks,
        "fee_status": "Pending",
        "created_at": datetime.now().isoformat()
    }
    
    # Initialize attendance
    db["attendance"][sid] = {}
    
    save_db(db)
    print(f"\n   ✅ Student added successfully! ID: {sid}")
    input("\n   Press Enter to continue...")

def view_students(db):
    """View all students with filtering and sorting"""
    print("\n" + "="*70)
    print("   👨‍🎓 STUDENT LIST")
    print("="*70)
    
    students = list(db["students"].values())
    
    if not students:
        print("\n   No students found!")
        input("\n   Press Enter to continue...")
        return
    
    # Filter options
    print("\n   Filter Options:")
    print("   1. Show All")
    print("   2. Filter by Class")
    print("   3. Search by Name")
    print("   4. Sort by Name")
    print("   5. Sort by Marks")
    print("   6. Sort by Attendance")
    
    choice = input("\n   Choice (1-6): ").strip()
    
    if choice == "2":
        cls = input("   Enter Class: ").strip()
        students = [s for s in students if s["class"] == cls]
    elif choice == "3":
        search = input("   Search Name: ").strip().lower()
        students = [s for s in students if search in s["name"].lower()]
    elif choice == "4":
        students.sort(key=lambda x: x["name"])
    elif choice == "5":
        students.sort(key=lambda x: sum(x["marks"].values()), reverse=True)
    elif choice == "6":
        students.sort(key=lambda x: get_attendance_percentage(db, x["id"]), reverse=True)
    
    # Display
    print(f"\n   {'ID':<10} {'Name':<20} {'Class':<10} {'Attendance':<12} {'Fee':<10}")
    print("   " + "-"*62)
    
    for s in students:
        att = get_attendance_percentage(db, s["id"])
        att_str = f"{att:.1f}%" if att > 0 else "N/A"
        low_att = " ⚠️" if 0 < att < 75 else ""
        print(f"   {s['id']:<10} {s['name']:<20} {s['class']}-{s['section']:<6} {att_str:<12}{low_att} {s.get('fee_status', 'Pending'):<10}")
    
    print(f"\n   Total Students: {len(students)}")
    input("\n   Press Enter to continue...")

def edit_student(db):
    """Edit student details"""
    print("\n" + "="*50)
    print("   ✏️ EDIT STUDENT")
    print("="*50)
    
    sid = input("\n   Enter Student ID: ").strip().upper()
    s = db["students"].get(sid)
    
    if not s:
        print("   ❌ Student not found!")
        input("\n   Press Enter to continue...")
        return
    
    print(f"\n   Current Details:")
    print(f"   Name: {s['name']}")
    print(f"   Class: {s['class']}-{s['section']}")
    print(f"   Father: {s.get('father_name', 'N/A')}")
    print(f"   Phone: {s.get('phone', 'N/A')}")
    
    print("\n   Enter new values (press Enter to keep current):")
    
    name = input(f"   Name [{s['name']}]: ").strip() or s['name']
    cls = input(f"   Class [{s['class']}]: ").strip() or s['class']
    section = input(f"   Section [{s['section']}]: ").strip() or s['section']
    father = input(f"   Father's Name [{s.get('father_name', '')}]: ").strip() or s.get('father_name', '')
    phone = input(f"   Phone [{s.get('phone', '')}]: ").strip() or s.get('phone', '')
    address = input(f"   Address [{s.get('address', '')}]: ").strip() or s.get('address', '')
    
    s["name"] = name
    s["class"] = cls
    s["section"] = section
    s["father_name"] = father
    s["phone"] = phone
    s["address"] = address
    
    save_db(db)
    print("\n   ✅ Student updated successfully!")
    input("\n   Press Enter to continue...")

def delete_student(db):
    """Delete a student"""
    print("\n" + "="*50)
    print("   🗑️ DELETE STUDENT")
    print("="*50)
    
    sid = input("\n   Enter Student ID: ").strip().upper()
    s = db["students"].get(sid)
    
    if not s:
        print("   ❌ Student not found!")
        input("\n   Press Enter to continue...")
        return
    
    print(f"\n   Student: {s['name']} ({sid})")
    confirm = input("   Are you sure you want to delete? (yes/no): ").strip().lower()
    
    if confirm == "yes":
        del db["students"][sid]
        db["attendance"].pop(sid, None)
        db["results"].pop(sid, None)
        save_db(db)
        print("   ✅ Student deleted!")
    else:
        print("   ❌ Deletion cancelled.")
    
    input("\n   Press Enter to continue...")

def view_student_profile(db):
    """View detailed student profile"""
    print("\n" + "="*50)
    print("   👤 STUDENT PROFILE")
    print("="*50)
    
    sid = input("\n   Enter Student ID: ").strip().upper()
    s = db["students"].get(sid)
    
    if not s:
        print("   ❌ Student not found!")
        input("\n   Press Enter to continue...")
        return
    
    # Calculate metrics
    marks = s.get("marks", {})
    vals = list(marks.values())
    total = sum(vals)
    avg = total / len(vals) if vals else 0
    grade = get_grade(avg)
    pass_status = "PASS" if all(v >= 35 for v in vals) and vals else "FAIL"
    att_pct = get_attendance_percentage(db, sid)
    
    print(f"""
    ┌────────────────────────────────────────┐
    │         STUDENT PROFILE CARD           │
    ├────────────────────────────────────────┤
    │ ID:        {s['id']:<27}│
    │ Name:      {s['name']:<27}│
    │ DOB:       {s.get('dob', 'N/A'):<27}│
    │ Class:     {s['class']}-{s['section']:<24}│
    │ Father:    {s.get('father_name', 'N/A'):<27}│
    │ Mother:    {s.get('mother_name', 'N/A'):<27}│
    │ Phone:     {s.get('phone', 'N/A'):<27}│
    │ Address:   {s.get('address', 'N/A'):<27}│
    ├────────────────────────────────────────┤
    │            ACADEMICS                   │
    ├────────────────────────────────────────┤""")
    
    for sub, mark in marks.items():
        print(f"    │ {sub:<20} {mark:>5.1f}                    │")
    
    print(f"""    ├────────────────────────────────────────┤
    │ Total:     {total:<27.1f}│
    │ Average:   {avg:<27.1f}│
    │ Grade:     {grade:<27}│
    │ Status:    {pass_status:<27}│
    ├────────────────────────────────────────┤
    │ Attendance:  {att_pct:.1f}%{'⚠️ LOW' if 0 < att_pct < 75 else '':>21}│
    │ Fee Status:  {s.get('fee_status', 'Pending'):<24}│
    └────────────────────────────────────────┘""")
    
    input("\n   Press Enter to continue...")

# ==================== MARKS MANAGEMENT ====================
def enter_marks(db):
    """Enter marks for a student"""
    print("\n" + "="*50)
    print("   📝 ENTER MARKS")
    print("="*50)
    
    sid = input("\n   Enter Student ID: ").strip().upper()
    s = db["students"].get(sid)
    
    if not s:
        print("   ❌ Student not found!")
        input("\n   Press Enter to continue...")
        return
    
    print(f"\n   Student: {s['name']} (Class {s['class']}-{s['section']})")
    
    # Exam type
    print("\n   Exam Types:")
    print("   1. Unit Test 1")
    print("   2. Unit Test 2")
    print("   3. Mid Term")
    print("   4. Final Exam")
    
    exam_choice = input("   Select Exam Type (1-4): ").strip()
    exam_types = {"1": "Unit Test 1", "2": "Unit Test 2", "3": "Mid Term", "4": "Final Exam"}
    exam_type = exam_types.get(exam_choice, "Final Exam")
    
    subjects = get_subjects_for_class(db, s["class"])
    
    print(f"\n   Enter marks for {exam_type}:")
    marks = {}
    
    for sub in subjects:
        current = s["marks"].get(sub, 0)
        mark_str = input(f"   {sub} (0-100) [{current}]: ").strip()
        marks[sub] = float(mark_str) if mark_str else current
    
    # Save to results
    if sid not in db["results"]:
        db["results"][sid] = {}
    db["results"][sid][exam_type] = marks
    
    # Update current marks
    db["students"][sid]["marks"] = marks
    
    save_db(db)
    print("\n   ✅ Marks saved successfully!")
    input("\n   Press Enter to continue...")

def view_results(db):
    """View results for all students"""
    print("\n" + "="*70)
    print("   📊 RESULTS REPORT")
    print("="*70)
    
    print("\n   Filter Options:")
    print("   1. All Students")
    print("   2. By Class")
    print("   3. Top Performers")
    
    choice = input("\n   Choice (1-3): ").strip()
    
    students = list(db["students"].values())
    
    if choice == "2":
        cls = input("   Enter Class: ").strip()
        students = [s for s in students if s["class"] == cls]
    elif choice == "3":
        students.sort(key=lambda x: sum(x["marks"].values()) if x["marks"] else 0, reverse=True)
        students = students[:10]
    
    print(f"\n   {'ID':<10} {'Name':<18} {'Total':<8} {'Avg':<8} {'Grade':<6} {'Status':<8}")
    print("   " + "-"*60)
    
    for s in students:
        marks = s.get("marks", {})
        vals = list(marks.values())
        total = sum(vals)
        avg = total / len(vals) if vals else 0
        grade = get_grade(avg)
        status = "PASS" if all(v >= 35 for v in vals) and vals else "FAIL"
        status_color = "✅" if status == "PASS" else "❌"
        
        print(f"   {s['id']:<10} {s['name']:<18} {total:<8.1f} {avg:<8.1f} {grade:<6} {status} {status_color}")
    
    input("\n   Press Enter to continue...")

# ==================== ATTENDANCE ====================
def mark_attendance(db):
    """Mark daily attendance"""
    print("\n" + "="*50)
    print("   ✅ MARK ATTENDANCE")
    print("="*50)
    
    date_str = input("\n   Date (YYYY-MM-DD) [Today]: ").strip()
    if not date_str:
        date_str = datetime.now().strftime("%Y-%m-%d")
    
    cls = input("   Class (blank for all): ").strip()
    
    students = list(db["students"].values())
    if cls:
        students = [s for s in students if s["class"] == cls]
    
    if not students:
        print("   No students found!")
        input("\n   Press Enter to continue...")
        return
    
    print(f"\n   Marking attendance for {date_str}")
    print("   (P = Present, A = Absent, S = Skip)")
    print("   " + "-"*40)
    
    for s in students:
        # Check if already marked
        current = db["attendance"].get(s["id"], {}).get(date_str)
        if current:
            print(f"   {s['id']} {s['name']}: Already marked as {current}")
            continue
        
        status = input(f"   {s['id']} {s['name']} [P/A]: ").strip().upper()
        
        if status == "S":
            continue
        
        if not db["attendance"].get(s["id"]):
            db["attendance"][s["id"]] = {}
        
        db["attendance"][s["id"]][date_str] = "Present" if status == "P" else "Absent"
    
    save_db(db)
    print("\n   ✅ Attendance marked successfully!")
    input("\n   Press Enter to continue...")

def bulk_attendance(db):
    """Mark bulk attendance"""
    print("\n" + "="*50)
    print("   📋 BULK ATTENDANCE")
    print("="*50)
    
    date_str = input("\n   Date (YYYY-MM-DD) [Today]: ").strip()
    if not date_str:
        date_str = datetime.now().strftime("%Y-%m-%d")
    
    cls = input("   Class (blank for all): ").strip()
    mark_all = input("   Mark all as (P/A): ").strip().upper()
    
    if mark_all not in ["P", "A"]:
        print("   ❌ Invalid choice!")
        return
    
    students = list(db["students"].values())
    if cls:
        students = [s for s in students if s["class"] == cls]
    
    status = "Present" if mark_all == "P" else "Absent"
    count = 0
    
    for s in students:
        if not db["attendance"].get(s["id"]):
            db["attendance"][s["id"]] = {}
        
        if date_str not in db["attendance"][s["id"]]:
            db["attendance"][s["id"]][date_str] = status
            count += 1
    
    save_db(db)
    print(f"\n   ✅ Marked {count} students as {status}!")
    input("\n   Press Enter to continue...")

def view_attendance_history(db):
    """View attendance history for a student"""
    print("\n" + "="*50)
    print("   📅 ATTENDANCE HISTORY")
    print("="*50)
    
    sid = input("\n   Enter Student ID: ").strip().upper()
    s = db["students"].get(sid)
    
    if not s:
        print("   ❌ Student not found!")
        input("\n   Press Enter to continue...")
        return
    
    print("\n   1. View All")
    print("   2. View by Month")
    print("   3. View by Date Range")
    
    choice = input("\n   Choice (1-3): ").strip()
    
    att = db["attendance"].get(sid, {})
    dates = sorted(att.keys())
    
    if choice == "2":
        month = input("   Enter month (YYYY-MM): ").strip()
        dates = [d for d in dates if d.startswith(month)]
    elif choice == "3":
        start = input("   Start date (YYYY-MM-DD): ").strip()
        end = input("   End date (YYYY-MM-DD): ").strip()
        dates = [d for d in dates if start <= d <= end]
    
    print(f"\n   Attendance for {s['name']} ({sid}):")
    print(f"   {'Date':<15} {'Status':<10}")
    print("   " + "-"*25)
    
    for d in dates:
        status = att[d]
        icon = "✅" if status == "Present" else "❌"
        print(f"   {d:<15} {status} {icon}")
    
    # Summary
    total_days = len(dates)
    present = sum(1 for d in dates if att[d] == "Present")
    absent = total_days - present
    pct = (present / total_days * 100) if total_days > 0 else 0
    
    print(f"\n   Summary: {present} Present, {absent} Absent, {pct:.1f}% attendance")
    
    if 0 < pct < 75:
        print("   ⚠️ LOW ATTENDANCE ALERT!")
    
    input("\n   Press Enter to continue...")

def attendance_report(db):
    """Generate attendance report for class"""
    print("\n" + "="*70)
    print("   📊 ATTENDANCE REPORT")
    print("="*70)
    
    cls = input("\n   Enter Class (blank for all): ").strip()
    
    students = list(db["students"].values())
    if cls:
        students = [s for s in students if s["class"] == cls]
    
    print(f"\n   {'ID':<10} {'Name':<18} {'Present':<10} {'Absent':<10} {'Percentage':<12} {'Status':<10}")
    print("   " + "-"*70)
    
    for s in students:
        att = db["attendance"].get(s["id"], {})
        total = len(att)
        present = sum(1 for v in att.values() if v == "Present")
        absent = total - present
        pct = (present / total * 100) if total > 0 else 0
        
        status = "✅ Good" if pct >= 75 else "⚠️ Low" if pct > 0 else "N/A"
        
        print(f"   {s['id']:<10} {s['name']:<18} {present:<10} {absent:<10} {pct:>6.1f}%     {status}")
    
    input("\n   Press Enter to continue...")

# ==================== FEE MANAGEMENT ====================
def manage_fees(db):
    """Manage fee structure"""
    print("\n" + "="*50)
    print("   💰 FEE STRUCTURE")
    print("="*50)
    
    print("\n   1. View Fee Structure")
    print("   2. Add/Edit Fee Structure")
    print("   3. Record Payment")
    print("   4. View Payments")
    print("   5. Fee Reports")
    
    choice = input("\n   Choice (1-5): ").strip()
    
    if choice == "1":
        view_fee_structure(db)
    elif choice == "2":
        edit_fee_structure(db)
    elif choice == "3":
        record_payment(db)
    elif choice == "4":
        view_payments(db)
    elif choice == "5":
        fee_reports(db)

def view_fee_structure(db):
    """View fee structure"""
    print("\n   Fee Structure:")
    print(f"   {'Class':<10} {'Admission':<12} {'Monthly':<12} {'Annual':<12}")
    print("   " + "-"*46)
    
    structure = db["fees"]["structure"]
    if not structure:
        print("   No fee structure set!")
    else:
        for cls, fees in sorted(structure.items()):
            print(f"   Class {cls:<5} ₹{fees['admission']:<11} ₹{fees['monthly']:<11} ₹{fees['annual']:<11}")

def edit_fee_structure(db):
    """Add or edit fee structure"""
    print("\n   Edit Fee Structure:")
    cls = input("   Class: ").strip()
    
    current = db["fees"]["structure"].get(cls, {})
    
    admission = input(f"   Admission Fee [₹{current.get('admission', 0)}]: ").strip()
    monthly = input(f"   Monthly Fee [₹{current.get('monthly', 0)}]: ").strip()
    annual = input(f"   Annual Fee [₹{current.get('annual', 0)}]: ").strip()
    
    db["fees"]["structure"][cls] = {
        "admission": float(admission) if admission else current.get("admission", 0),
        "monthly": float(monthly) if monthly else current.get("monthly", 0),
        "annual": float(annual) if annual else current.get("annual", 0)
    }
    
    save_db(db)
    print("\n   ✅ Fee structure saved!")

def record_payment(db):
    """Record a fee payment"""
    print("\n   Record Fee Payment:")
    
    sid = input("   Student ID: ").strip().upper()
    s = db["students"].get(sid)
    
    if not s:
        print("   ❌ Student not found!")
        return
    
    print(f"\n   Student: {s['name']} (Class {s['class']})")
    
    print("\n   Fee Types:")
    print("   1. Admission Fee")
    print("   2. Monthly Fee")
    print("   3. Annual Fee")
    print("   4. Custom Amount")
    
    type_choice = input("   Select type (1-4): ").strip()
    fee_types = {"1": "Admission", "2": "Monthly", "3": "Annual", "4": "Custom"}
    fee_type = fee_types.get(type_choice, "Custom")
    
    # Get suggested amount
    structure = db["fees"]["structure"].get(s["class"], {})
    suggested = structure.get(fee_type.lower(), 0) if fee_type != "Custom" else 0
    
    amount_str = input(f"   Amount [₹{suggested}]: ").strip()
    amount = float(amount_str) if amount_str else suggested
    
    period = input("   Period (e.g., January 2024): ").strip()
    
    receipt_no = f"RCP-{uuid.uuid4().hex[:8].upper()}"
    
    db["fees"]["payments"].append({
        "date": datetime.now().isoformat(),
        "student_id": sid,
        "name": s["name"],
        "class": s["class"],
        "type": fee_type,
        "amount": amount,
        "period": period,
        "receipt_no": receipt_no
    })
    
    # Update student fee status
    db["students"][sid]["fee_status"] = "Paid"
    
    save_db(db)
    print(f"\n   ✅ Payment recorded! Receipt: {receipt_no}")

def view_payments(db):
    """View payment history"""
    print("\n   Payment History:")
    print(f"   {'Date':<12} {'ID':<10} {'Name':<18} {'Type':<12} {'Amount':<10} {'Receipt'}")
    print("   " + "-"*72)
    
    for p in reversed(db["fees"]["payments"]):
        date = p["date"][:10]
        print(f"   {date:<12} {p['student_id']:<10} {p['name']:<18} {p['type']:<12} ₹{p['amount']:<9} {p['receipt_no']}")

def fee_reports(db):
    """Generate fee reports"""
    print("\n   Fee Report:")
    
    total_collected = sum(p["amount"] for p in db["fees"]["payments"])
    total_students = len(db["students"])
    paid_students = len(set(p["student_id"] for p in db["fees"]["payments"]))
    
    print(f"\n   Total Collection: ₹{total_collected:,.2f}")
    print(f"   Total Students: {total_students}")
    print(f"   Students with Payments: {paid_students}")
    print(f"   Students Pending: {total_students - paid_students}")
    
    # By class
    class_wise = defaultdict(float)
    for p in db["fees"]["payments"]:
        class_wise[p.get("class", "N/A")] += p["amount"]
    
    print("\n   Class-wise Collection:")
    for cls, amount in sorted(class_wise.items()):
        print(f"   Class {cls}: ₹{amount:,.2f}")

# ==================== TIMETABLE ====================
def manage_timetable(db):
    """Manage class timetable"""
    print("\n" + "="*50)
    print("   📅 CLASS TIMETABLE")
    print("="*50)
    
    print("\n   1. View Timetable")
    print("   2. Edit Timetable")
    
    choice = input("\n   Choice (1-2): ").strip()
    
    if choice == "1":
        view_timetable(db)
    else:
        edit_timetable(db)

def view_timetable(db):
    """View class timetable"""
    cls = input("\n   Enter Class: ").strip()
    section = input("   Enter Section (A/B/C): ").strip() or "A"
    
    key = f"{cls}-{section}"
    timetable = db["timetable"].get(key, {})
    
    days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"]
    periods = ["1", "2", "3", "4", "5", "6"]
    times = ["9:00-10:00", "10:00-11:00", "11:15-12:15", "1:00-2:00", "2:00-3:00", "3:00-4:00"]
    
    print(f"\n   Timetable for Class {cls}-{section}:")
    print(f"   {'Time':<15}", end="")
    for day in days:
        print(f"{day:<12}", end="")
    print()
    print("   " + "-"*87)
    
    for i, period in enumerate(periods):
        print(f"   {times[i]:<15}", end="")
        for day in days:
            subject = timetable.get(f"{day}-{period}", "-")
            print(f"{subject:<12}", end="")
        print()

def edit_timetable(db):
    """Edit class timetable"""
    cls = input("\n   Enter Class: ").strip()
    section = input("   Enter Section (A/B/C): ").strip() or "A"
    
    key = f"{cls}-{section}"
    if key not in db["timetable"]:
        db["timetable"][key] = {}
    
    subjects = get_subjects_for_class(db, cls)
    
    print("\n   Available subjects:")
    for i, sub in enumerate(subjects, 1):
        print(f"   {i}. {sub}")
    
    days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"]
    periods = ["1", "2", "3", "4", "5", "6"]
    
    print("\n   Enter subject number for each slot (0 to skip):")
    
    for day in days:
        print(f"\n   {day}:")
        for period in periods:
            choice = input(f"   Period {period}: ").strip()
            if choice and choice != "0":
                try:
                    idx = int(choice) - 1
                    if 0 <= idx < len(subjects):
                        db["timetable"][key][f"{day}-{period}"] = subjects[idx]
                except ValueError:
                    pass
    
    save_db(db)
    print("\n   ✅ Timetable updated!")

# ==================== NOTICES ====================
def manage_notices(db):
    """Manage notice board"""
    print("\n" + "="*50)
    print("   📢 NOTICE BOARD")
    print("="*50)
    
    print("\n   1. View Notices")
    print("   2. Add Notice")
    print("   3. Delete Notice")
    
    choice = input("\n   Choice (1-3): ").strip()
    
    if choice == "1":
        view_notices(db)
    elif choice == "2":
        add_notice(db)
    elif choice == "3":
        delete_notice(db)

def view_notices(db):
    """View all notices"""
    notices = db["notices"]
    
    if not notices:
        print("\n   No notices!")
        return
    
    print("\n   Notices:")
    print("   " + "-"*50)
    
    for i, n in enumerate(reversed(notices), 1):
        priority_icon = "🔴" if n["priority"] == "urgent" else "🟡" if n["priority"] == "important" else "🟢"
        print(f"\n   {i}. {priority_icon} {n['title']}")
        print(f"      Date: {n['date'][:10]}")
        print(f"      {n['content']}")

def add_notice(db):
    """Add a new notice"""
    title = input("\n   Notice Title: ").strip()
    content = input("   Notice Content: ").strip()
    
    print("\n   Priority:")
    print("   1. Normal")
    print("   2. Important")
    print("   3. Urgent")
    
    priority_choice = input("   Select (1-3): ").strip()
    priorities = {"1": "normal", "2": "important", "3": "urgent"}
    priority = priorities.get(priority_choice, "normal")
    
    db["notices"].append({
        "id": str(uuid.uuid4()),
        "title": title,
        "content": content,
        "priority": priority,
        "date": datetime.now().isoformat()
    })
    
    save_db(db)
    print("\n   ✅ Notice published!")

def delete_notice(db):
    """Delete a notice"""
    view_notices(db)
    
    if not db["notices"]:
        return
    
    idx = int(input("\n   Enter notice number to delete: ").strip())
    
    if 1 <= idx <= len(db["notices"]):
        db["notices"].pop(-idx)
        save_db(db)
        print("   ✅ Notice deleted!")

# ==================== ID CARD ====================
def generate_id_card(db):
    """Generate student ID card"""
    print("\n" + "="*50)
    print("   🪪 ID CARD GENERATOR")
    print("="*50)
    
    sid = input("\n   Enter Student ID: ").strip().upper()
    s = db["students"].get(sid)
    
    if not s:
        print("   ❌ Student not found!")
        input("\n   Press Enter to continue...")
        return
    
    school_name = input("   School Name [Bright Future School]: ").strip() or "Bright Future School"
    session = input("   Academic Session [2024-2025]: ").strip() or "2024-2025"
    
    print(f"""
    ╔══════════════════════════════════════════════════╗
    ║                                                  ║
    ║           {school_name:^36}║
    ║         Academic Session: {session:^22}║
    ║                                                  ║
    ╠══════════════════════════════════════════════════╣
    ║                                                  ║
    ║              STUDENT ID CARD                     ║
    ║                                                  ║
    ║   Name:     {s['name']:<35}║
    ║   ID No:    {s['id']:<35}║
    ║   Class:    {s['class']}-{s['section']:<33}║
    ║   DOB:      {s.get('dob', 'N/A'):<35}║
    ║   Father:   {s.get('father_name', 'N/A'):<35}║
    ║   Phone:    {s.get('phone', 'N/A'):<35}║
    ║   Address:  {s.get('address', 'N/A')[:30]:<35}║
    ║                                                  ║
    ╠══════════════════════════════════════════════════╣
    ║                                                  ║
    ║  ________________        ________________         ║
    ║  Principal                Parent                  ║
    ║                                                  ║
    ╚══════════════════════════════════════════════════╝""")
    
    input("\n   Press Enter to continue...")

# ==================== SUBJECTS ====================
def manage_subjects(db):
    """Manage subjects"""
    print("\n" + "="*50)
    print("   📚 SUBJECT MANAGEMENT")
    print("="*50)
    
    print("\n   1. View Subjects")
    print("   2. Add Global Subject")
    print("   3. Add Class-specific Subject")
    print("   4. Remove Subject")
    
    choice = input("\n   Choice (1-4): ").strip()
    
    if choice == "1":
        view_subjects(db)
    elif choice == "2":
        add_global_subject(db)
    elif choice == "3":
        add_class_subject(db)
    elif choice == "4":
        remove_subject(db)

def view_subjects(db):
    """View all subjects"""
    print("\n   Global Subjects (All Classes):")
    for i, sub in enumerate(db["subjects"]["global"], 1):
        print(f"   {i}. {sub}")
    
    print("\n   Class-wise Subjects:")
    for cls, subjects in db["subjects"]["classwise"].items():
        print(f"\n   Class {cls}:")
        for sub in subjects:
            print(f"     - {sub}")

def add_global_subject(db):
    """Add a global subject"""
    name = input("\n   Subject Name: ").strip()
    if name and name not in db["subjects"]["global"]:
        db["subjects"]["global"].append(name)
        save_db(db)
        print(f"   ✅ '{name}' added to global subjects!")
    else:
        print("   ❌ Subject already exists or invalid name!")

def add_class_subject(db):
    """Add a class-specific subject"""
    cls = input("\n   Class: ").strip()
    name = input("   Subject Name: ").strip()
    
    if cls not in db["subjects"]["classwise"]:
        db["subjects"]["classwise"][cls] = []
    
    if name and name not in db["subjects"]["classwise"][cls]:
        db["subjects"]["classwise"][cls].append(name)
        save_db(db)
        print(f"   ✅ '{name}' added to Class {cls}!")
    else:
        print("   ❌ Subject already exists or invalid name!")

def remove_subject(db):
    """Remove a subject"""
    print("\n   Remove from:")
    print("   1. Global Subjects")
    print("   2. Class-specific Subjects")
    
    choice = input("\n   Choice (1-2): ").strip()
    
    if choice == "1":
        view_subjects(db)
        name = input("\n   Enter subject name to remove: ").strip()
        if name in db["subjects"]["global"]:
            db["subjects"]["global"].remove(name)
            save_db(db)
            print(f"   ✅ '{name}' removed!")
    elif choice == "2":
        cls = input("\n   Class: ").strip()
        if cls in db["subjects"]["classwise"]:
            print(f"\n   Class {cls} subjects:")
            for sub in db["subjects"]["classwise"][cls]:
                print(f"     - {sub}")
            
            name = input("\n   Enter subject name to remove: ").strip()
            if name in db["subjects"]["classwise"][cls]:
                db["subjects"]["classwise"][cls].remove(name)
                save_db(db)
                print(f"   ✅ '{name}' removed!")

# ==================== ANALYTICS ====================
def view_analytics(db):
    """View analytics and reports"""
    print("\n" + "="*70)
    print("   📊 ANALYTICS & REPORTS")
    print("="*70)
    
    students = list(db["students"].values())
    total = len(students)
    
    # Class distribution
    class_dist = defaultdict(int)
    for s in students:
        class_dist[s["class"]] += 1
    
    print(f"\n   📈 Overview:")
    print(f"   Total Students: {total}")
    print(f"   Total Notices: {len(db['notices'])}")
    print(f"   Total Fee Payments: {len(db['fees']['payments'])}")
    total_collection = sum(p["amount"] for p in db["fees"]["payments"])
    print(f"   Total Fee Collection: ₹{total_collection:,.2f}")
    
    # Class distribution
    print(f"\n   📊 Class Distribution:")
    for cls in sorted(class_dist.keys()):
        bar = "█" * class_dist[cls]
        print(f"   Class {cls:>2}: {bar} ({class_dist[cls]})")
    
    # Attendance summary
    today = datetime.now().strftime("%Y-%m-%d")
    present_today = sum(1 for sid in db["attendance"] if db["attendance"][sid].get(today) == "Present")
    
    print(f"\n   ✅ Attendance Today: {present_today}/{total}")
    
    # Low attendance alerts
    low_att_count = 0
    for s in students:
        att_pct = get_attendance_percentage(db, s["id"])
        if 0 < att_pct < 75:
            low_att_count += 1
    
    print(f"   ⚠️ Low Attendance Alerts: {low_att_count}")
    
    # Grade distribution
    grades = {"A+": 0, "A": 0, "B": 0, "C": 0, "D": 0, "F": 0}
    for s in students:
        marks = s.get("marks", {})
        vals = list(marks.values())
        if vals:
            avg = sum(vals) / len(vals)
            grade = get_grade(avg)
            grades[grade] += 1
    
    print(f"\n   🎓 Grade Distribution:")
    for grade, count in grades.items():
        bar = "█" * count
        print(f"   {grade:>2}: {bar} ({count})")
    
    # Top performers
    print(f"\n   🏆 Top 5 Performers:")
    performers = []
    for s in students:
        marks = s.get("marks", {})
        vals = list(marks.values())
        if vals:
            avg = sum(vals) / len(vals)
            performers.append((s["id"], s["name"], avg))
    
    performers.sort(key=lambda x: x[2], reverse=True)
    
    for i, (sid, name, avg) in enumerate(performers[:5], 1):
        grade = get_grade(avg)
        print(f"   {i}. {name} ({sid}) - {avg:.1f}% [{grade}]")
    
    input("\n   Press Enter to continue...")

# ==================== DATA IMPORT/EXPORT ====================
def export_data(db):
    """Export data to file"""
    print("\n   Export Data:")
    print("   1. Export as JSON")
    print("   2. Export as CSV")
    
    choice = input("\n   Choice (1-2): ").strip()
    
    if choice == "1":
        filename = f"school_backup_{datetime.now().strftime('%Y%m%d_%H%M%S')}.json"
        with open(filename, "w") as f:
            json.dump(db, f, indent=2)
        print(f"\n   ✅ Data exported to {filename}")
    
    elif choice == "2":
        filename = f"students_{datetime.now().strftime('%Y%m%d')}.csv"
        with open(filename, "w", newline="") as f:
            writer = csv.writer(f)
            writer.writerow(["ID", "Name", "Class", "Section", "Father", "Phone", "Address", "Fee Status"])
            for s in db["students"].values():
                writer.writerow([
                    s["id"], s["name"], s["class"], s["section"],
                    s.get("father_name", ""), s.get("phone", ""),
                    s.get("address", ""), s.get("fee_status", "Pending")
                ])
        print(f"\n   ✅ Data exported to {filename}")

def import_data(db):
    """Import data from file"""
    print("\n   Import Data:")
    print("   1. Import JSON")
    print("   2. Import CSV")
    
    choice = input("\n   Choice (1-2): ").strip()
    filename = input("   Enter filename: ").strip()
    
    if not os.path.exists(filename):
        print("   ❌ File not found!")
        return
    
    try:
        if choice == "1":
            with open(filename, "r") as f:
                imported = json.load(f)
            
            if "students" in imported:
                db["students"].update(imported["students"])
                db["counter"] = max(db["counter"], imported.get("counter", 0))
            
            if "attendance" in imported:
                for sid, att in imported["attendance"].items():
                    if sid not in db["attendance"]:
                        db["attendance"][sid] = {}
                    db["attendance"][sid].update(att)
            
            save_db(db)
            print(f"\n   ✅ Data imported successfully!")
        
        elif choice == "2":
            with open(filename, "r") as f:
                reader = csv.DictReader(f)
                count = 0
                for row in reader:
                    sid = generate_student_id(db)
                    db["students"][sid] = {
                        "id": sid,
                        "name": row.get("Name", ""),
                        "class": row.get("Class", "1"),
                        "section": row.get("Section", "A"),
                        "father_name": row.get("Father", ""),
                        "phone": row.get("Phone", ""),
                        "address": row.get("Address", ""),
                        "marks": {},
                        "fee_status": "Pending"
                    }
                    db["attendance"][sid] = {}
                    count += 1
            
            save_db(db)
            print(f"\n   ✅ Imported {count} students!")
    
    except Exception as e:
        print(f"   ❌ Error: {e}")

# ==================== MAIN MENU ====================
def main_menu(db):
    """Display and handle main menu"""
    while True:
        print("\n" + "="*60)
        print("   🏫 SCHOOL MANAGEMENT SYSTEM v2.0")
        print("="*60)
        print("""
   1.  👨‍🎓 Student Management
   2.  📝 Marks & Results
   3.  ✅ Attendance Management
   4.  💰 Fee Management
   5.  📅 Timetable
   6.  📢 Notice Board
   7.  🪪 ID Card Generator
   8.  📚 Subject Management
   9.  📊 Analytics & Reports
   10. 📁 Data Import/Export
   11. 🔐 Change PIN
   0.  🚪 Logout
        """)
        
        choice = input("   Enter choice (0-11): ").strip()
        
        if choice == "1":
            student_menu(db)
        elif choice == "2":
            marks_menu(db)
        elif choice == "3":
            attendance_menu(db)
        elif choice == "4":
            manage_fees(db)
        elif choice == "5":
            manage_timetable(db)
        elif choice == "6":
            manage_notices(db)
        elif choice == "7":
            generate_id_card(db)
        elif choice == "8":
            manage_subjects(db)
        elif choice == "9":
            view_analytics(db)
        elif choice == "10":
            data_menu(db)
        elif choice == "11":
            change_pin(db)
        elif choice == "0":
            save_db(db)
            print("\n   👋 Thank you! Goodbye!")
            break

def student_menu(db):
    """Student management submenu"""
    while True:
        print("\n" + "-"*40)
        print("   👨‍🎓 STUDENT MANAGEMENT")
        print("-"*40)
        print("""
   1. Add Student
   2. View All Students
   3. View Student Profile
   4. Edit Student
   5. Delete Student
   0. Back to Main Menu
        """)
        
        choice = input("   Choice: ").strip()
        
        if choice == "1": add_student(db)
        elif choice == "2": view_students(db)
        elif choice == "3": view_student_profile(db)
        elif choice == "4": edit_student(db)
        elif choice == "5": delete_student(db)
        elif choice == "0": break

def marks_menu(db):
    """Marks management submenu"""
    while True:
        print("\n" + "-"*40)
        print("   📝 MARKS & RESULTS")
        print("-"*40)
        print("""
   1. Enter Marks
   2. View Results
   0. Back to Main Menu
        """)
        
        choice = input("   Choice: ").strip()
        
        if choice == "1": enter_marks(db)
        elif choice == "2": view_results(db)
        elif choice == "0": break

def attendance_menu(db):
    """Attendance management submenu"""
    while True:
        print("\n" + "-"*40)
        print("   ✅ ATTENDANCE MANAGEMENT")
        print("-"*40)
        print("""
   1. Mark Daily Attendance
   2. Bulk Attendance
   3. View Attendance History
   4. Attendance Report
   0. Back to Main Menu
        """)
        
        choice = input("   Choice: ").strip()
        
        if choice == "1": mark_attendance(db)
        elif choice == "2": bulk_attendance(db)
        elif choice == "3": view_attendance_history(db)
        elif choice == "4": attendance_report(db)
        elif choice == "0": break

def data_menu(db):
    """Data import/export submenu"""
    while True:
        print("\n" + "-"*40)
        print("   📁 DATA IMPORT/EXPORT")
        print("-"*40)
        print("""
   1. Export Data
   2. Import Data
   0. Back to Main Menu
        """)
        
        choice = input("   Choice: ").strip()
        
        if choice == "1": export_data(db)
        elif choice == "2": import_data(db)
        elif choice == "0": break

# ==================== MAIN ====================
def main():
    """Main function"""
    print("\n" + "╔" + "═"*58 + "╗")
    print("║" + " "*15 + "🏫 SCHOOL MANAGEMENT SYSTEM" + " "*15 + "║")
    print("║" + " "*20 + "Version 2.0" + " "*26 + "║")
    print("╚" + "═"*58 + "╝")
    
    db = load_db()
    
    if not login(db):
        return
    
    main_menu(db)

if __name__ == "__main__":
    main()
```
